reading a json file recursively

```python
import json

def read_json_file(file_path):
    with open(file_path, 'r') as file:
        data = json.load(file)
    return data

def recursive_read(data, indent=0):
  if isinstance(data,dict):
    for key, value in data.items():
      print(''*indent + str(key)+':') #So " " * indent creates indentation depending on how deep we are in the JSON structure.
      recursive_read(value,indent+2)
  elif isinstance(data,list):
    for index, item in enumerate(data):
      print(''*indent + f"[{index}]")
      recursive_read(item, indent + 2)

  else:
    print(""* indent + str(data))







```
